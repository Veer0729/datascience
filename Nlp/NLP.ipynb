```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("train.txt", sep= ";", header = None, names = ["text", "emotion"])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
unique_emotions = df["emotion"].unique()
emotion_numbers = {}
i = 0

for emo in unique_emotions:
  emotion_numbers[emo] = i
  i += 1

df["emotion"] = df["emotion"].map(emotion_numbers)
```

```python
df
```

```python
df["text"] = df["text"].apply(lambda x : x.lower()) # lowercasing
```

```python
import string
def remove_punc(txt):
  return txt.translate(str.maketrans("","",string.punctuation))
```

```python
df["text"].apply(remove_punc)
```

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

```python
def remove_emojis(txt):
  new = ""
  for i in txt:
    if not i.isascii():
      new += i
  return new
  df["text"] = df["text"].apply(remove_emojis)
```

```python
print(df["text"].head())
```

```python
# NLTK: Natural lang toolkit
import nltk

# A collection of text is called a Corpus
# A group of words forming a maningful statement is called a sentence
# Each individual word or symbol in the text is called a token
```

```python
print(df["text"].head())
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("stopwords")
```

```python
stop_words = set(stopwords.words("english"))
stop_words
```

```python
def remove(txt):
    words = str(txt).split()
    cleaned = []

    for i in words:
        if not i in stop_words:
            cleaned.append(i)

    return ' '.join(cleaned)
```

```python
df["text"] = df["text"].apply(remove)
```

```python
## BAG OF WORDS vectorization technique

from sklearn.feature_extraction.text import CountVectorizer

documents = [
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta is great"
]

vectorizer = CountVectorizer(ngram_range=(2,2)) # Bigrams in sklearn

X = vectorizer.fit_transform(documents)

print("Vocabulary:", vectorizer.get_feature_names_out())
print("\nBoW Matrix:\n", X.toarray())
```

```python
df.head()
```

```python
# This splits your data: 80% for training the model, 20% held back to test how well it generalizes

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['emotion'], test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
# CountVectorizer turns each sentence into a vector of word counts (Bag of Words)..

from sklearn.naive_bayes import MultinomialNB
# MultinomialNB is a Naive Bayes classifier well-suited for word count data.
# Trains on the BoW vectors, predicts emotions for the test set.
from sklearn.metrics import accuracy_score

bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
# .fit_transform() on training data: learns the vocabulary AND converts the text.

X_test_bow = bow_vectorizer.transform(X_test)
# .transform() on test data: reuses the same vocabulary.


nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)


pred_bow = nb_model.predict(X_test_bow)
print(accuracy_score(y_test, pred_bow))
```

```python
# TfidfVectorizer is like CountVectorizer, but instead of raw counts, it weighs words by how distinctive they are
  #(common words like "the" get down-weighted; rare, meaningful words get boosted)

tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)


nb2_model = MultinomialNB()
# Same Naive Bayes model, but trained on TF-IDF vectors instead.

nb2_model.fit(X_train_tfidf,y_train)
```

```python
y_pred = nb2_model.predict(X_test_tfidf)
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(max_iter=1000)
# max_iter=1000 increases the number of optimization iterations so the model has enough room to converge

logistic_model.fit(X_train_tfidf, y_train)
log_pred = logistic_model.predict(X_test_tfidf)

print(accuracy_score(y_test, log_pred))
```
