```python
import numpy as np # working with arrays, maths
import pandas as pd # works with tables
import matplotlib.pyplot as plt
import seaborn as sns # this and matplotlib help in visualisation
```

```python
df = pd.read_csv("Diwali Sales Data.csv", encoding= "unicode_escape") # to avoid encoding error
df.shape # the total rows and columns of our data set
```

```python
df.head(10) # top 5 data
```

```python
df.info()
```

```python
df.drop(["Status", "unnamed1"], axis = 1, inplace = True)

# drop for deleting particular columns
# axis = 1 means deleting the whole column
# inplace = true means to save the changes
```

```python
pd.isnull(df) # check null values
```

```python
pd.isnull(df).sum()
```

```python
df.dropna(inplace=True) # drops null values
```

```python
df["Amount"] = df["Amount"].astype("int") # to change data type
df
```

```python
df.columns # to see all our colums
```

```python
df.rename(columns= {"Marital_Status":"Shaadi"}) # to change name...it wont be saved as we didnt use inplace = True
```

```python
df.describe() # tells us count, mean, median, mode, std
```

```python
df[["Age", "Orders", "Amount"]].describe()
```

# EDA

```python
# GENDER
```

```python
gen = sns.countplot(x = "Gender", data= df)

for bars in gen.containers:
    gen.bar_label(bars)
```

```python
df.groupby(["Gender"], as_index= False)["Amount"].sum().sort_values(by = "Amount", ascending=False)
```

```python
sales_gen = df.groupby(["Gender"], as_index= False)["Amount"].sum().sort_values(by = "Amount", ascending=False)
sns.barplot(x = "Gender", y = "Amount", data = sales_gen)
```

```python
# AGE
```

```python
ax = sns.countplot(data = df, x = "Age Group", hue = "Gender")

for bars in ax.containers:
    ax.bar_label(bars)

# hue means comparison
```

```python
# Total amount vs Age group
sales_age = df.groupby(['Age Group'], as_index=False)["Amount"].sum().sort_values(by = "Amount", ascending=False)

sns.barplot(x = "Age Group", y = "Amount", data = sales_age)
```

```python
# STATE
```

```python
sales_state = df.groupby(['State'], as_index = False)['Amount'].sum().sort_values(by = "Amount", ascending = False).head()

sns.set(rc = {'figure.figsize': (18, 5)})
sns.barplot(data = sales_state, x = "State", y = "Amount")
```

```python
sales_state = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by = "Orders", ascending = False).head()

sns.set(rc = {'figure.figsize': (18, 5)})
sns.barplot(data = sales_state, x = "State", y = "Orders")
```

```python
# MARTIAL STATUS
```

```python
ax = sns.countplot(data = df, x = "Marital_Status")

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_mar = df.groupby(['Marital_Status', "Gender"], as_index= False)["Amount"].sum().sort_values(by = "Amount", ascending= False)

sns.set(rc ={'figure.figsize' : (6,5)})
sns.barplot(data = sales_mar, x = "Marital_Status", y = "Amount", hue = "Gender")
```

most buyers are married women

```python
# OCCUPATION
```

```python
sns.set(rc = {"figure.figsize":(20, 5)})
ax = sns.countplot(data = df, x = "Occupation")

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_mar = df.groupby(['Occupation'], as_index= False)["Amount"].sum().sort_values(by = "Amount", ascending= False)

sns.set(rc ={'figure.figsize' : (15,5)})
sns.barplot(data = sales_mar, x = "Occupation", y = "Amount")
```

```python
# PRODUCT CATEGORY
```

```python
sns.set(rc = {"figure.figsize":(20, 5)})
ax = sns.countplot(data = df, x = "Product_Category")

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_mar = df.groupby(['Product_Category'], as_index= False)["Amount"].sum().sort_values(by = "Amount", ascending= False)

sns.set(rc ={'figure.figsize' : (15,5)})
sns.barplot(data = sales_mar, x = "Product_Category", y = "Amount")
```

```python
sales_state = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_ID',y= 'Orders')
```

```python
# top 10 most sold products (same thing as above)

fig1, ax1 = plt.subplots(figsize=(12,7))
df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending=False).plot(kind='bar')
```
